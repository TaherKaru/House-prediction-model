```python
df= df[~df['price'].str.contains(r"-|\|",na=False)].copy()
df.shape
```

```python
def convert_price(values):
    if pd.isna(values):
        return np.nan
    
    values = str(values).strip().lower()
    values = values.replace("₹", "").replace(",", "").strip()
    
    if "cr" in values:
        values = values.replace("cr", "").strip()
        return float(values) * 10_000_000
    
    elif "lac" in values or "lakh" in values or "l" in values:
        values = (
            values.replace("lakh", "")
                  .replace("lac", "")
                  .replace("l", "")
                  .strip()
        )
        return float(values) * 100_000
    
    try:
        return float(values)
    except ValueError:
        return np.nan
    
df["price"] = df["price"].apply(convert_price)
```

```python
def convert_price_per_sqft(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()
    value = value.replace("₹", "").replace(",", "").strip()

    if "l" in value and "/sqm" in value:
        value = value.replace("l", "").replace("/sqm", "").strip()
        return float(value) * 100_000

    elif "l" in value and "/sqft" in value:
        value = value.replace("l", "").replace("/sqft", "").strip()
        return float(value) * 100_000

    elif "/sqft" in value:
        value = value.replace("/sqft", "").strip()
        return float(value)

    elif "/sqm" in value:
        value = value.replace("/sqm", "").strip()
        return float(value)

    return np.nan

df["price_per_sqft"] = df["price_per_sqft"].apply(convert_price_per_sqft)
```

```python
def convert_area(value):
    if pd.isna(value):
        return np.nan

    value = str(value).lower().strip()

    value = value.replace(",", "")

    if "sqm" in value:
        value = (
            value.replace("sqm", "")
                 .replace("(", "")
                 .replace(")", "")
                 .strip()
        )
        return float(value) * 10.7639

    elif "sqft" in value:
        value = (
            value.replace("sqft", "")
                 .replace("(", "")
                 .replace(")", "")
                 .strip()
        )
        return float(value)

    elif "/" in value:
        value = value.replace("/", "").strip()
        return float(value)

    try:
        return float(value)

    except ValueError:
        return np.nan


df["area"] = df["area"].apply(convert_area)


df["area"] = df["area"].apply(convert_area)
```

```python
def convert_area_unit(value):
    if pd.isna(value):
        return np.nan

    value = str(value).lower().strip()

    value = (
        value.replace(",", "")
             .replace("(", "")
             .replace(")", "")
             .strip()
    )

    if "sqm" in value:
        value = value.replace("sqm", "").strip()
        return float(value) * 10.7639

    elif "sqft" in value:
        value = value.replace("sqft", "").strip()
        return float(value)

    try:
        return float(value)

    except ValueError:
        return np.nan


df["area_unit"] = df["area_unit"].apply(convert_area_unit)
```
