```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_mumbai_house_data.csv")

df.head()
df.info()
df.describe()
```

# 1.univariate analysis


```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'], bins=30, kde=True)
plt.title("Price Distribution")
plt.show()
```

```python
print("Skewness :", df["price"].skew())
print("Kurtosis:", df["price"].kurtosis())
```

```python
df["log_price"] = np.log1p(df["price"])
print(df["log_price"].skew())
print(df["log_price"].kurtosis())
```

```python
sns.histplot(
    df["log_price"],
    bins=30,
    kde=True
)
```

## Conclusion

- The price variable is highly right-skewed and has a very heavy tail, as shown by the high skewness (6.14) and kurtosis (67.86).
- This suggests that a few very high-priced houses are strongly affecting the distribution.
- After applying a log transformation, the distribution becomes much more normal, with skewness reduced to about 0.40 and kurtosis to about 0.04.
- Therefore, log-transformed price is better suited for modeling and further statistical analysis than the original price column.
